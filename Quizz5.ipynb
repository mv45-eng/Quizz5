```python
#-------TRES EN RAYA
class Jugador:
    def __init__(self, nombre, simbolo):
        self.nombre = nombre
        self.simbolo = simbolo

    def hacer_jugada(self, tablero):
        while True:
            try:
                posicion = int(input(f"{self.nombre} ({self.simbolo}), elige una casilla (1-9): "))
                if posicion in tablero and tablero[posicion] == " ":
                    tablero[posicion] = self.simbolo
                    break
                else:
                    print("Casilla inválida o ya ocupada.")
            except:
                print("Entrada inválida. Ingresa un número del 1 al 9.")


class Juego:
    def __init__(self):
        self.tablero = {i: " " for i in range(1, 10)}
        self.jugador1 = Jugador("Jugador 1", "X")
        self.jugador2 = Jugador("Jugador 2", "O")
        self.turno_actual = self.jugador1

    def mostrar_tablero(self):
        t = self.tablero
        print(f"""
 {t[1]} | {t[2]} | {t[3]}
-----------
 {t[4]} | {t[5]} | {t[6]}
-----------
 {t[7]} | {t[8]} | {t[9]}
""")

    def verificar_ganador(self):
        combinaciones = [
            (1,2,3), (4,5,6), (7,8,9),
            (1,4,7), (2,5,8), (3,6,9),
            (1,5,9), (3,5,7)
        ]

        for a, b, c in combinaciones:
            if self.tablero[a] == self.tablero[b] == self.tablero[c] != " ":
                return True
        return False

    def tablero_lleno(self):
        return all(valor != " " for valor in self.tablero.values())

    def cambiar_turno(self):
        if self.turno_actual == self.jugador1:
            self.turno_actual = self.jugador2
        else:
            self.turno_actual = self.jugador1

    def jugar(self):
        while True:
            self.mostrar_tablero()
            self.turno_actual.hacer_jugada(self.tablero)

            if self.verificar_ganador():
                self.mostrar_tablero()
                print(f"{self.turno_actual.nombre} gana.")
                break

            if self.tablero_lleno():
                self.mostrar_tablero()
                print("Empate.")
                break

            self.cambiar_turno()

juego = Juego()
juego.jugar()
```
